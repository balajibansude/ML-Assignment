**Assignment 5**Take input of hours spent on different activities over a week and visualize the data with pie
charts using matplotlib to show percentage time distribution

1. Aim

The aim of this assignment is to take input of the hours spent on various activities over the course of a week and visualize this data using pie charts with the help of the matplotlib library in Python. The objective is to show the percentage distribution of time spent on different activities and provide insights into how time is allocated across various tasks.

2. Theory

Data Visualization:
Data visualization is a graphical representation of data and information. By using visual elements like charts, graphs, and maps, data visualization tools provide an accessible way to see and understand trends, outliers, and patterns in data.

Pie Chart:
A pie chart is a circular statistical graphic, divided into slices to illustrate numerical proportions. In this assignment, pie charts will be used to visualize the percentage of time spent on different activities over the week. Each slice of the pie chart represents the time spent on a particular activity relative to the total time.

Matplotlib:
Matplotlib is a widely-used Python library for creating static, animated, and interactive visualizations. It is commonly used for creating line plots, bar charts, histograms, and pie charts, among other types of visualizations. In this assignment, the matplotlib.pyplot module will be used to create pie charts to display the data.

3. Program Flow / Algorithm

Input Collection:

Prompt the user to input the number of hours spent on various activities over the course of the week.

The user will input the hours for a set of predefined activities (e.g., sleep, work, exercise, study, entertainment).

Data Processing:

Store the inputted hours in a list or dictionary.

Calculate the total number of hours spent across all activities.

Data Visualization:

Use matplotlib.pyplot to create a pie chart that shows the percentage distribution of time spent on different activities.

Output:

Display the pie chart with proper labels and percentages.

Conclusion:

Based on the pie chart, make conclusions about time management and suggest areas where improvements can be made if necessary.

```python
# Importing necessary library
import matplotlib.pyplot as plt

# Step 1: Input number of activities
num_activities = int(input("Enter the number of activities: "))

# Step 2: Input activity names and hours spent
activities = []
hours = []

for i in range(num_activities):
    activity_name = input(f"Enter name of activity {i+1}: ")
    activity_hours = float(input(f"Enter hours spent on {activity_name} in a week: "))

    activities.append(activity_name)
    hours.append(activity_hours)

# Step 3: Plotting pie chart
plt.figure(figsize=(8,8))
plt.pie(hours, labels=activities, autopct='%1.1f%%', startangle=140)
plt.title("Percentage of Time Spent on Activities in a Week")
plt.show()
```

**Conclusion**
In this assignment, we successfully visualized the distribution of time spent on various activities over the course of a week using a pie chart. By analyzing the pie chart, we can quickly identify which activities take up the most time and which are the least prioritized. This kind of visualization can be very useful in time management, allowing individuals to better allocate their time across different tasks.

For instance, if the "Work" or "Study" sections take up an excessively large portion of time, an individual might decide to allocate more time for rest or exercise. Visualization tools like pie charts make these insights clearer and easier to interpret.
